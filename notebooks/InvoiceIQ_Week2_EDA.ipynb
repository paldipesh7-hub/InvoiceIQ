```python
# Select 20,000 records from each currency

boxplot_sample = (
    df.groupby("currency", group_keys=False)
      .sample(n=20000, random_state=42)
)

print("Original dataset:", len(df))
print("Boxplot sample:", len(boxplot_sample))

print("\nSample count by currency:")
print(boxplot_sample["currency"].value_counts())
```

```python
# Create six boxplots, one for each currency

fig, axes = plt.subplots(2, 3, figsize=(15, 9))

currencies = sorted(boxplot_sample["currency"].unique())

for ax, currency in zip(axes.flatten(), currencies):

    currency_data = boxplot_sample.loc[
        boxplot_sample["currency"] == currency,
        "total_amount"
    ]

    ax.boxplot(
        currency_data,
        showfliers=False
    )

    ax.set_title(f"Invoice Amount - {currency}")
    ax.set_ylabel("Invoice Amount")
    ax.set_xticks([])

# Add an overall title
fig.suptitle(
    "Boxplot of Invoice Amounts by Currency",
    fontsize=16
)

plt.tight_layout()
plt.show()
```

```python
# Calculate quartiles using the full dataset

amount_quartiles = (
    df.groupby("currency")["total_amount"]
      .agg(
          minimum="min",
          Q1=lambda x: x.quantile(0.25),
          median="median",
          Q3=lambda x: x.quantile(0.75),
          maximum="max"
      )
)

# Calculate the interquartile range
amount_quartiles["IQR"] = (
    amount_quartiles["Q3"] - amount_quartiles["Q1"]
)

print(amount_quartiles.round(2))
```

```python
# Identify the day of the month for every invoice
df["invoice_day"] = df["invoice_date"].dt.day

# Identify the final five days of each month
df["is_month_end"] = (
    df["invoice_date"].dt.day >=
    (df["invoice_date"].dt.days_in_month - 4)
)

# Count invoices in each group
month_end_analysis = (
    df.groupby("is_month_end")
    .agg(
        invoice_count=("invoice_id", "count"),
        average_invoice_amount=("total_amount", "mean")
    )
    .reset_index()
)

month_end_analysis["period"] = month_end_analysis[
    "is_month_end"
].map({
    True: "Last 5 Days of Month",
    False: "Other Days of Month"
})

print(month_end_analysis)
```

```python
plt.figure(figsize=(10, 6))

sns.barplot(
    data=month_end_analysis,
    x="period",
    y="invoice_count"
)

plt.title("Invoice Volume: Month-End vs Other Days", fontsize=15)
plt.xlabel("Invoice Period")
plt.ylabel("Number of Invoices")
plt.xticks(rotation=10)
plt.tight_layout()
plt.show()
```

```python
# Find the final date of each invoice's calendar quarter
quarter_end_date = (
    df["invoice_date"]
    .dt.to_period("Q")
    .dt.end_time
    .dt.normalize()
)

# Identify invoices issued in the final five days of the quarter
df["is_quarter_end"] = (
    df["invoice_date"] >= quarter_end_date - pd.Timedelta(days=4)
)

quarter_end_analysis = (
    df.groupby("is_quarter_end")
    .agg(
        invoice_count=("invoice_id", "count"),
        average_invoice_amount=("total_amount", "mean")
    )
    .reset_index()
)

quarter_end_analysis["period"] = quarter_end_analysis[
    "is_quarter_end"
].map({
    True: "Last 5 Days of Quarter",
    False: "Other Days of Quarter"
})

print(quarter_end_analysis)
```

```python
plt.figure(figsize=(10, 6))

sns.barplot(
    data=quarter_end_analysis,
    x="period",
    y="invoice_count"
)

plt.title("Invoice Volume: Quarter-End vs Other Days", fontsize=15)
plt.xlabel("Invoice Period")
plt.ylabel("Number of Invoices")
plt.xticks(rotation=10)
plt.tight_layout()
plt.show()
```

```python
# Create a calendar containing every date in the dataset
calendar_dates = pd.date_range(
    start=df["invoice_date"].min(),
    end=df["invoice_date"].max(),
    freq="D"
)

calendar_df = pd.DataFrame({
    "invoice_date": calendar_dates
})

# Identify the last 5 days of each month
calendar_df["is_month_end"] = (
    calendar_df["invoice_date"].dt.day
    >= calendar_df["invoice_date"].dt.days_in_month - 4
)

# Identify the last 5 days of each quarter
quarter_end_dates = (
    calendar_df["invoice_date"]
    .dt.to_period("Q")
    .dt.end_time
    .dt.normalize()
)

calendar_df["is_quarter_end"] = (
    calendar_df["invoice_date"]
    >= quarter_end_dates - pd.Timedelta(days=4)
)

# Count the actual calendar days in each group
month_days = (
    calendar_df.groupby("is_month_end")
    .size()
    .reset_index(name="calendar_days")
)

quarter_days = (
    calendar_df.groupby("is_quarter_end")
    .size()
    .reset_index(name="calendar_days")
)

# Merge calendar days with invoice counts
month_end_daily = month_end_analysis.merge(
    month_days, on="is_month_end"
)

quarter_end_daily = quarter_end_analysis.merge(
    quarter_days, on="is_quarter_end"
)

# Calculate average invoices per calendar day
month_end_daily["invoices_per_day"] = (
    month_end_daily["invoice_count"]
    / month_end_daily["calendar_days"]
)

quarter_end_daily["invoices_per_day"] = (
    quarter_end_daily["invoice_count"]
    / quarter_end_daily["calendar_days"]
)

print("MONTH-END DAILY COMPARISON")
print(month_end_daily[[
    "period", "invoice_count", "calendar_days", "invoices_per_day"
]])

print("\nQUARTER-END DAILY COMPARISON")
print(quarter_end_daily[[
    "period", "invoice_count", "calendar_days", "invoices_per_day"
]])
```

```python
# Group outstanding invoices by currency
outstanding_amount_by_currency = (
    outstanding_invoices
    .groupby("currency")
    .agg(
        outstanding_invoice_count=("invoice_id", "count"),
        outstanding_amount=("total_amount", "sum"),
        average_invoice_amount=("total_amount", "mean")
    )
    .reset_index()
)

# Round monetary values to 2 decimal places
outstanding_amount_by_currency[
    ["outstanding_amount", "average_invoice_amount"]
] = outstanding_amount_by_currency[
    ["outstanding_amount", "average_invoice_amount"]
].round(2)

# Display the results
print("OUTSTANDING AMOUNT BY CURRENCY")
print("--------------------------------")

display(outstanding_amount_by_currency)
```

```python
# Define the reporting cutoff date
as_of_date = pd.Timestamp("2025-12-31")

# Identify outstanding invoices past their due dates
past_due_invoices = outstanding_invoices[
    outstanding_invoices["due_date"] < as_of_date
].copy()

# Calculate number of past-due invoices
past_due_count = len(past_due_invoices)

# Calculate the number of outstanding invoices not yet past due
not_yet_due_count = (
    len(outstanding_invoices) - past_due_count
)

# Display results
print("PAST-DUE INVOICE ANALYSIS")
print("------------------------------")

print("Reporting date:", as_of_date.date())

print("Total outstanding invoices:", len(outstanding_invoices))

print("Past-due invoices:", past_due_count)

print("Not-yet-due outstanding invoices:", not_yet_due_count)

print(
    "Past-due percentage:",
    round(past_due_count / len(outstanding_invoices) * 100, 2),
    "%"
)
```

```python
# Calculate the number of days each invoice is past due
past_due_invoices["days_past_due"] = (
    as_of_date - past_due_invoices["due_date"]
).dt.days

# Define the aging bucket labels
aging_labels = [
    "0-30 days",
    "31-60 days",
    "61-90 days",
    "90+ days"
]

# Assign each invoice to an aging bucket
past_due_invoices["aging_bucket"] = pd.cut(
    past_due_invoices["days_past_due"],
    bins=[0, 30, 60, 90, float("inf")],
    labels=aging_labels,
    right=True
)

# Count invoices in each aging bucket
aging_summary = (
    past_due_invoices["aging_bucket"]
    .value_counts(sort=False)
    .reset_index()
)

aging_summary.columns = ["aging_bucket", "invoice_count"]

# Calculate percentage of past-due invoices
aging_summary["percentage"] = (
    aging_summary["invoice_count"] /
    len(past_due_invoices) * 100
).round(2)

# Display results
print("INVOICE AGING ANALYSIS")
print("------------------------------")

display(aging_summary)
```

```python
# Calculate outstanding amounts by aging bucket and currency
aging_amount_summary = (
    past_due_invoices
    .groupby(
        ["currency", "aging_bucket"],
        observed=False
    )
    .agg(
        invoice_count=("invoice_id", "count"),
        outstanding_amount=("total_amount", "sum")
    )
    .reset_index()
)

# Round outstanding amounts
aging_amount_summary["outstanding_amount"] = (
    aging_amount_summary["outstanding_amount"].round(2)
)

print("OUTSTANDING AMOUNT BY AGING BUCKET AND CURRENCY")
print("------------------------------------------------")

display(aging_amount_summary)
```

### Business Interpretation

The dataset indicates predominantly early payment behaviour. However, the absence of verified late payments means that actual late-payment behaviour cannot be studied using this dataset.

A future model could investigate factors associated with early versus on-time payment. However, a genuine late-payment prediction model requires additional representative data containing verified late-payment examples.
