```python
# Create a separate cleaned copy
df_clean = df.copy()

# 1. Clean text columns
text_columns = df_clean.select_dtypes(
    include=["object", "string"]
).columns

for column in text_columns:
    df_clean[column] = df_clean[column].str.strip()

# 2. Standardise currency and payment status
df_clean["currency"] = df_clean["currency"].str.upper()
df_clean["payment_status"] = df_clean["payment_status"].str.title()

# 3. Keep payment dates only for Paid invoices
df_clean["payment_date_clean"] = df_clean["payment_date"].where(
    df_clean["payment_status"] == "Paid"
)

# 4. Calculate invoice-to-due days
df_clean["invoice_to_due_days"] = (
    df_clean["due_date"] - df_clean["invoice_date"]
).dt.days

# 5. Calculate days to pay
df_clean["days_to_pay"] = (
    df_clean["payment_date_clean"] - df_clean["invoice_date"]
).dt.days

# 6. Calculate payment delay
df_clean["payment_delay_days"] = (
    df_clean["payment_date_clean"] - df_clean["due_date"]
).dt.days

# 7. Create paid-late flag
df_clean["paid_late_flag"] = pd.Series(
    pd.NA, index=df_clean.index, dtype="Int8"
)

paid_mask = df_clean["payment_status"] == "Paid"

df_clean.loc[paid_mask, "paid_late_flag"] = (
    df_clean.loc[paid_mask, "payment_delay_days"] > 0
).astype("int8")

# 8. Create payment indicators
df_clean["is_paid"] = (
    df_clean["payment_status"] == "Paid"
).astype("int8")

df_clean["is_open"] = (
    df_clean["payment_status"].isin(["Pending", "Overdue"])
).astype("int8")

# 9. Create overdue target
df_clean["overdue_target"] = (
    df_clean["payment_status"] == "Overdue"
).astype("int8")

print("Cleaning completed successfully!")
print("Dataset shape:", df_clean.shape)
```

```python
print("Dataset shape:")
print(df_clean.shape)

print("\nMissing values:")
print(df_clean.isnull().sum().sort_values(ascending=False).head(15))

print("\nPayment status:")
print(df_clean["payment_status"].value_counts())

print("\nPaid late flag:")
print(df_clean["paid_late_flag"].value_counts(dropna=False))

print("\nPayment delay statistics:")
print(df_clean.loc[paid_mask, "payment_delay_days"].describe())
```
